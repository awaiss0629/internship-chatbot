```python
from getpass import getpass
from pyngrok import ngrok, conf

# Clean start
ngrok.kill()

# (Optional) force v3 agent; pyngrok will download it if missing
conf.get_default().ngrok_version = "v3"

token = getpass("Paste your ngrok authtoken: ")
ngrok.set_auth_token(token)     # writes a valid config & installs agent if needed
print("✅ ngrok authtoken set")
```
